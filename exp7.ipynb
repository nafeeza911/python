```python
# colors from colorset 1 not contained in color set2. 
color_set1 = set(input("Enter colors for set1 seperated by commas: ").split(","))
color_set2 = set(input("Enter colors for set2 seperated by commas: ").split(","))
result = color_set1 - color_set2
print("Colors in set 1 but not in set 2:", result)
```

```python
#printout all colors from color set 1 and color set2
set1 = set(input("Enter Color Set 1: ").split(","))
set2 = set(input("Enter Color Set 2: ").split(","))
print("set1: ",set1)
print("set2: ",set2)
print("set1 and set2 combined: ",set1 | set2)
```
