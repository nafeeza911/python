```python
# Program to display first and last colors
colors = input("Enter colors: ").split(",")
print("First color:", colors[0])
print("Last color:", colors[-1])
```

```python
#prints extension of a file
filename = input("Enter the filename: ")
if "." in filename:
    extension = filename.rsplit(".", 1)[1]
    print("File extension:", extension)
else:
    print("The file has no extension.")
```

```python
# Program to find sum of numbers
numbers = input("Enter numbers: ")
num_list = [int(x) for x in numbers.split(",")]
total = sum(num_list)
print("List:", num_list)
print("Sum:", total)
```

```python
# Program to find sum of numbers
numbers = input("Enter numbers seperated by commas: ")
num_list = [int(x) for x in numbers.split(",")]
total = sum(num_list)
print("List:", num_list)
print("Sum:", total)
```
