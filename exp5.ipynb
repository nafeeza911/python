```python
#get the 3rd element and 3rd element from last of a tuple.
my_tuple = (10, 20, 30, 40, 50, 60, 70)
print("3rd element:", my_tuple[2])
print("3rd element from last:", my_tuple[-3])
```

```python
#get the 3rd element and 3rd element from last of a tuple.
t = tuple(map(int, input("Enter tuple elements separated by space: ").split()))
print("3rd element:", t[2])
print("3rd element from last:", t[-3])
```

```python
#generate a tuple and find average of all items in that tuple.
numbers = input("Enter comma-separated numbers: ")
t = tuple(map(int, numbers.split(",")))
average = sum(t) / len(t)
print("Tuple:", t)
print("Average:", average)
```
