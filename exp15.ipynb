```python
#Rectangle class with attributes length and breadth and methods to find area and perimeter.
class Rectangle:
    def get(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        return self.l * self.b

    def perimeter(self):
        return 2 * (self.l + self.b)


r1 = Rectangle()
r2 = Rectangle()

l1 = int(input("Enter the length of 1st rectangle: "))
b1 = int(input("Enter the breadth of 1st rectangle: "))
l2 = int(input("Enter the length of 2nd rectangle: "))
b2 = int(input("Enter the breadth of 2nd rectangle: "))
 
r1.get(l1,b1)
r2.get(l2,b2)

print("Area of Rectangle 1:",r1.area())
print("Area of Rectangle 2:",r2.area())
print("perimeter of rectangle 1:",r1.perimeter())
print("perimeter of Rectangle 2:",r2.perimeter())

if r1.area() > r2.area():
    print("Rectangle 1 has a greater area.")
elif r1.area() < r2.area():
    print("Rectangle 2 has a greater area.")
else:
    print("Both rectangles have equal area.")
```

```python
#Bank Account Using Constructor, Deposit and Withdrawal Methods
class bank:
    def __init__(self,acc_no,name,type,balance):
        self.acc_no=acc_no
        self.name=name
        self.type=type
        self.balance=balance
    def deposit(self,amount):
        self.balance=self.balance+amount
        print("Deposited amount:",amount)
        print("updated balance:",self.balance)
    def withdraw(self,amount):
        print("Balance=",self.balance)
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print("Withdrawn amount:", amount)
            print("Updated balance:", self.balance)

    def display(self):
        print("Account Number:",self.acc_no)
        print("Name:",self.name)
        print("Account type:",self.type)
        print("Balance:",self.balance)
        
acc_no=int(input("Enter account number:"))
name=input("Enter the name:")
type=input("Account type is:")
balance=float(input("Enter the initial balance:"))


account=bank(acc_no,name,type,balance)

while True:
    print("1.Deposit")
    print("2.Withdraw")
    print("3.Display")
    print("4.Exit")
    choice=int(input("Enter your choice:"))
    if choice==1:
        amt=int(input("Enter amount to deposit:"))
        account.deposit(amt)
    elif choice==2:
        amt=int(input("Enter amount to withdraw:"))
        account.withdraw(amt)
        
    elif choice==3:
        account.display()
    elif choice==4:
        print("Thankyou!")
        break
    else:
        print("Invalid choice.Try again!")
```

```python
#Program Using Multilevel Inheritance, Constructor Invocation and Method Overriding
class publisher:
    def __init__(self,name):
        self.name=name
        
    def display(self):
        print("Publisher name:",self.name)
        
class book(publisher):
    def __init__(self,name,title,author):
        super().__init__(name)
        self.title=title
        self.author=author
    def display(self):
        super().display()
        print("book title:",self.title)
        print("author name:",self.author)

class python(book):
    def __init__(self,name,title,author,price,pages_no):
        super().__init__(name, title, author)
        self.price=price
        self.pages_no=pages_no
        
    def display(self):
        super().display()
        print("Book Price:",self.price)
        print("Number of pages:",self.pages_no)

name=input("Enter publisher name:")
title=input("Enter the book title:")
author=input("Enter author's name:")
price=input("Enter the price:")
pages_no=int(input("Enter the number of pages:"))

print("\n")
book=python(name,title,author,price,pages_no)
book.display()
```

```python
#Rectangle Area Comparison Using Operator Overloading
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def __lt__(self, other):
        return self.area() < other.area()


l1 = int(input("Enter length of Rectangle 1: "))
w1 = int(input("Enter width of Rectangle 1: "))

l2 = int(input("Enter length of Rectangle 2: "))
w2 = int(input("Enter width of Rectangle 2: "))

r1 = Rectangle(l1, w1)
r2 = Rectangle(l2, w2)

print("Area of Rectangle 1:", r1.area())
print("Area of Rectangle 2:", r2.area())

if r1 < r2:
    print("Rectangle 1 has smaller area.")
elif r2 < r1:
    print("Rectangle 2 has smaller area.")
else:
    print("Both rectangles have equal area.")
```
