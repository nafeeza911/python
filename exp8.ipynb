```python
#Python program to accept 5 students' names and roll numbers, and map these into a dictionary
students = {}
students[int(input("Enter roll number 1: "))] = input("Enter name 1: ")
students[int(input("Enter roll number 2: "))] = input("Enter name 2: ")
students[int(input("Enter roll number 3: "))] = input("Enter name 3: ")
students[int(input("Enter roll number 4: "))] = input("Enter name 4: ")
students[int(input("Enter roll number 5: "))] = input("Enter name 5: ")
print("Original Dictionary:", students)

ascending = dict(sorted(students.items()))
print("Ascending Order:", ascending)

descending = dict(sorted(students.items(), reverse=True))
print("Descending Order:", descending)
```

```python
# Merge 2 dictionaries and display it
dic1 = {}

dic1[int(input("Enter roll number 1: "))] = input("Enter name 1: ")
dic1[int(input("Enter roll number 2: "))] = input("Enter name 2: ")
dic1[int(input("Enter roll number 3: "))] = input("Enter name 3: ")

dic2 = {}

dic2[int(input("Enter roll number 4: "))] = input("Enter name 4: ")
dic2[int(input("Enter roll number 5: "))] = input("Enter name 5: ")

dic1.update(dic2)

print("Merged Dictionary:", dic1)
```
