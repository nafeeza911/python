```python
# Program to exchange the first and last characters of a string
string = input("Enter a string: ")
if len(string) <= 1:
    new_string = string
else:
    new_string = string[-1] + string[1:-1] + string[0]
print("New string:", new_string)
```

```python
# Program to get the first 2 and last 2 characters of a string
string = input("Enter a string: ")
new_string = string[:2] + string[-2:]
print("New string:", new_string)
```

```python
#create a single string separated with space from two strings by swapping the character at position
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")
result = s2[0] + s1[1:] + " " + s1[0] + s2[1:]
print(result)
```

```python
#ccurrences of its first char have been changed to $, except the first char itself
s = input("Enter a string: ")
result = s[0] + s[1:].replace(s[0], "$")
print(result)
```
