```python
#count the occurrences of each word in a line of text.
text = input("Enter a line: ")
words = text.split()
count = {}
for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

print(count)
```

```python
#all values greater than 100, store ‘over’ instead.
numbers = list(map(int, input("Enter numbers: ").split()))
for i in range(len(numbers)):
    if numbers[i] > 100:
        numbers[i] = "over"
print(numbers)
```

```python
#Count the occurrences of ‘a’ within the list
names = input("Enter names: ").split()
print("List:", names)

count = 0

for name in names:
    count += name.lower().count("a")
print("Number of 'a':", count)
```

```python
#gcd of 2 numbers
a = int(input("enter first number: "))
b = int(input("enter second number: "))
gcd = 1
for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i
print("gcd: ", gcd)

```

```python
#ist removing even numbers
numbers = list(map(int, input("enter numbers: ").split()))
list2 = []

for num in numbers:
    if num % 2 != 0:
        list2.append(num)
print("List after removing even numbers:",list2)
```

```python
#factorial of a number
n = int(input("enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i
print("factorial is: ", fact)
```

```python
#Fibonacci series of N terms
n = int(input("enter number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a,b = b,a + b
```

```python
#sum of all items in a list
numbers = list(map(int, input("Enter numbers: ").split()))
print("list =", numbers)

total = 0

for num in numbers:
    total = total + num
print("Sum of numbers in list:", total)
```

```python
#list of four digit numbers in a given range with all their digits even and the number is a perfect square.
start = int(input("enter the starting four digit number:"))
end = int(input("enter the ending four digit number: "))
for num in range(start, end + 1):
    if all(int(digit) % 2 == 0 for digit in str(num)):
        root = int(num ** 0.5)

        if root * root == num:
            print(num)
```

```python
#Count the number of characters in a string
text = input("enter a string: ")

for char in set(text):
    print(char,":",text.count(char))
```

```python
#Check list length, list sum, and common values
list1 = list(map(int, input("Enter first list: ").split()))
list2 = list(map(int, input("Enter second list: ").split()))

print("Same length:", len(list1) == len(list2))

print("Same sum:", sum(list1) == sum(list2))

print("Common value:", bool(set(list1) & set(list2)))

```

```python
#pyramid with step number
n = int(input("Enter N: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i * j, end=" ")
    print()
```
